```python
import math

class Balok:
    # Variabel Statis (Class Variables)
    jumVolumeTotalBalok = 0.0
    jumBalok = 0
    Volume = 0.0
    luasPermukaan = 0.0
    diagonalRuang = 0.0
    digSisiDepan = 0.0
    digSisiSamping = 0.0
    digSisiAtas = 0.0

    def __init__(self):
        # Variabel Instance (Panjang, Lebar, Tinggi)
        self.panjang = 0.0
        self.lebar = 0.0
        self.tinggi = 0.0

    def hitungVolume(self):
        Balok.Volume = self.panjang * self.lebar * self.tinggi
        return Balok.Volume

    def hitungLuasPermukaan(self):
        Balok.luasPermukaan = 2 * (self.panjang * self.lebar + self.panjang * self.tinggi + self.tinggi * self.lebar)
        return Balok.luasPermukaan

    def hitungDiagonalRuang(self):
        Balok.diagonalRuang = math.sqrt((self.hitungDiagonalSisiDepan() ** 2) + (self.lebar ** 2))
        return Balok.diagonalRuang

    def hitungDiagonalSisiDepan(self):
        Balok.digSisiDepan = math.sqrt((self.panjang ** 2) + (self.tinggi ** 2))
        return Balok.digSisiDepan

    # Di Python, access modifier (protected/private) biasanya ditandai dengan garis bawah (underscore).
    # Namun agar panggilan dari luar class tetap mudah seperti di Java, kita biarkan namanya standar.
    def hitungDiagonalSisiSamping(self):
        Balok.digSisiSamping = math.sqrt((self.lebar ** 2) + (self.tinggi ** 2))
        return Balok.digSisiSamping

    def hitungDiagonalSisiAtas(self):
        Balok.digSisiAtas = math.sqrt((self.panjang ** 2) + (self.lebar ** 2))
        return Balok.digSisiAtas

    @classmethod
    def jumTotalVolume(cls):
        cls.jumVolumeTotalBalok = cls.jumVolumeTotalBalok + cls.Volume
        return cls.jumVolumeTotalBalok

    @classmethod
    def tamJumlahBalok(cls):
        cls.jumBalok += 1

    def tampilData(self):
        print(f"Panjang               : {self.panjang}")
        print(f"Lebar                 : {self.lebar}")
        print(f"Tinggi                : {self.tinggi}")


# Blok Main
if __name__ == "__main__":
    b1 = Balok()
    b1.panjang = 4
    b1.lebar = 2
    b1.tinggi = 3

    b1.hitungVolume()
    b1.hitungLuasPermukaan()
    b1.hitungDiagonalSisiDepan()
    b1.hitungDiagonalRuang()
    b1.hitungDiagonalSisiSamping()
    b1.hitungDiagonalSisiAtas()
    Balok.jumTotalVolume()
    Balok.tamJumlahBalok()

    b1.tampilData()
    print(f"Volume                : {b1.hitungVolume()}")
    print(f"Luas Permukaan        : {b1.hitungLuasPermukaan()}")
    print(f"Diagonal Ruang        : {b1.hitungDiagonalRuang()}")
    print(f"Diagonal Sisi Depan   : {b1.hitungDiagonalSisiDepan()}")
    print(f"Diagonal Sisi Samping : {b1.hitungDiagonalSisiSamping()}")
    print(f"Diagonal Sisi Atas    : {b1.hitungDiagonalSisiAtas()}")

    print("\n------------------------------------\n")

    b2 = Balok()
    b2.panjang = 5
    b2.lebar = 3
    b2.tinggi = 4

    b2.hitungVolume()
    b2.hitungLuasPermukaan()
    b2.hitungDiagonalSisiDepan()
    b2.hitungDiagonalRuang()
    b2.hitungDiagonalSisiSamping()
    b2.hitungDiagonalSisiAtas()
    Balok.tamJumlahBalok()

    # Sesuai urutan di Java, hitung volume masuk ke total sebelum print terakhir
    # (Perhatikan di Java Anda memanggil jumTotalVolume di dalam fungsi print di bawah)

    b2.tampilData()
    print(f"Volume                : {b2.hitungVolume()}")
    print(f"Luas Permukaan        : {b2.hitungLuasPermukaan()}")
    print(f"Diagonal Ruang        : {b2.hitungDiagonalRuang()}")
    print(f"Diagonal Sisi Depan   : {b2.hitungDiagonalSisiDepan()}")
    print(f"Diagonal Sisi Samping : {b2.hitungDiagonalSisiSamping()}")
    print(f"Diagonal Sisi Atas    : {b2.hitungDiagonalSisiAtas()}")
    print(f"JUMLAH SEMUA VOLUME BALOK    : {Balok.jumTotalVolume()}")
    print(f"JUMLAH BALOK YANG DIBUAT     : {Balok.jumBalok}")
```
